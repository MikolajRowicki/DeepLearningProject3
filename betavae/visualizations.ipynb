```python
import matplotlib.pyplot as plt
import numpy as np

labels = ['Cats Only', 'Cats & Dogs']
means = [1.61, 2.11] 
stds = [0.05, 0.01]   

plt.figure(figsize=(6, 5))
plt.style.use('seaborn-v0_8-whitegrid') 

x_pos = np.arange(len(labels))
bars = plt.bar(x_pos, means, yerr=stds, align='center', alpha=0.85, 
               color=['#4C72B0', '#DD8452'], capsize=8, edgecolor='black', linewidth=1.2)

plt.ylabel('Mean FID (Lower is better)', fontsize=12, fontweight='bold')
plt.title(r'$\beta$-VAE: Cats vs. Cats & Dogs (200 Epochs)', fontsize=14, pad=15)
plt.xticks(x_pos, labels, fontsize=12)
for bar, mean, std in zip(bars, means, stds):
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + std + 0.03, 
             f'{mean:.2f} ± {std:.2f}', ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.ylim(0, 2.5)

plt.tight_layout()
plt.savefig('vae_cats_vs_catsdogs.png', dpi=300)
plt.show()
```
